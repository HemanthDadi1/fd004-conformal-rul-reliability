# Paper 1 five-seed UQ aggregator

Run this **after** the IID, DS-A and DS-B confirmatory notebooks. It combines the six method/condition summaries and creates a manuscript-ready mean ± SD table. It does not train anything.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python

import os, pandas as pd
base = "/content/drive/MyDrive/FD004_Research/Paper1_MultiSeed_UQ"
if not os.path.isdir(base):
    base = "/content/Paper1_MultiSeed_UQ"

files = [
    "Paper1_FiveSeed_IID_per_seed.csv",
    "Paper1_FiveSeed_DS_A_per_seed.csv",
    "Paper1_FiveSeed_DS_B_per_seed.csv",
]
missing = [f for f in files if not os.path.exists(os.path.join(base,f))]
if missing:
    raise FileNotFoundError("Missing outputs: " + ", ".join(missing))

all_seed = pd.concat([pd.read_csv(os.path.join(base,f)) for f in files], ignore_index=True)

metrics = ["PICP","MPIW","Winkler","RMSE","MAE"]
rows = []
for (condition,method),g in all_seed.groupby(["Condition","Method"],sort=False):
    row = {"Condition":condition,"Method":method,"N_Seeds":len(g)}
    for m in metrics:
        row[m+"_mean"] = g[m].mean()
        row[m+"_sd"] = g[m].std(ddof=1)
        row[m+"_mean±sd"] = f"{g[m].mean():.2f} ± {g[m].std(ddof=1):.2f}"
    rows.append(row)

summary = pd.DataFrame(rows)
summary.to_csv(os.path.join(base,"Paper1_FiveSeed_UQ_ALL_summary.csv"),index=False)
all_seed.to_csv(os.path.join(base,"Paper1_FiveSeed_UQ_ALL_per_seed.csv"),index=False)

display(summary[[
    "Condition","Method","N_Seeds",
    "PICP_mean±sd","MPIW_mean±sd","Winkler_mean±sd","RMSE_mean±sd","MAE_mean±sd"
]])
print("Saved combined outputs to", base)
```
